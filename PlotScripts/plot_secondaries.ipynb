```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
```

```python
df = pd.read_csv("Secondary.csv")
#print(df["name"].value_counts())
#print(df.columns)
```

```python
for particle in df["name"].unique():
    df[df["name"]==particle]["ekin_MeV"].hist(bins=50, alpha=0.5, label=particle)
plt.xlabel("Energía [MeV]")
plt.ylabel("Cuenta")
plt.title("Distribución de energías por tipo de partícula")
plt.legend()
plt.show()
```

```python
fig, axes = plt.subplots(1, 3, figsize=(15, 4))  # 1 fila, 3 columnas

# ux
axes[0].hist(df["momX"], bins=20)
axes[0].set_xlabel('Coseno director en x')
axes[0].set_ylabel('Número de partículas')
axes[0].set_title('Histograma ux')

# uy
axes[1].hist(df["momX"], bins=20)
axes[1].set_xlabel('Coseno director en y')
axes[1].set_ylabel('Número de partículas')
axes[1].set_title('Histograma uy')

# uz
axes[2].hist(df["momX"], bins=20)
axes[2].set_xlabel('Coseno director en z')
axes[2].set_ylabel('Número de partículas')
axes[2].set_title('Histograma uz')
fig.suptitle("Cosenos Directores del momento", fontsize=16)
plt.tight_layout()
plt.show()
```

```python
plt.subplot(1, 2, 1)
plt.scatter(df["posX_mm"], df["posY_mm"], alpha=0.5)
plt.xlabel("X [mm]")
plt.ylabel("Y [mm]")
plt.title("Posiciones XY de generación")
plt.axis('equal')

plt.subplot(1, 2, 2)
plt.scatter(df["posZ_mm"], df["posX_mm"], alpha=0.5)
plt.xlabel("X [mm]")
plt.ylabel("Z [mm]")
plt.title("Posiciones XZ de generación")
plt.axis('equal')
```

```python
df["name"].value_counts().plot(kind='bar')
plt.xlabel("Tipo de partícula")
plt.ylabel("Cuenta")
plt.title("Número de partículas por tipo")
plt.show()
```
